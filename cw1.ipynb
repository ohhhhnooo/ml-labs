# Данные

Для последующих заданий будем использовать обезличенные транзакционные банковские данные. Для этого считайте в переменные **transactions_short, tr_mcc_codes и gender_train** из одноимённых таблиц из папки data.

###  Описание данных
#### Таблица ```transactions_short.csv```
##### Описание
Таблица содержит историю транзакций клиентов банка за один год и три месяца.

##### Формат данных

```
customer_id,tr_datetime,mcc_code,tr_type,amount,term_id
111111,15 01:40:52,1111,1000,-5224,111111
111112,15 15:18:32,3333,2000,-100,11122233
...
```
##### Описание полей

 - ```customer_id``` — идентификатор клиента;
 - ```tr_datetime``` — день и время совершения транзакции (дни нумеруются с начала данных);
 - ```mcc_code``` — mcc-код транзакции;
 - ```tr_type``` — тип транзакции;
 - ```amount``` — сумма транзакции в условных единицах со знаком; ```+``` — начисление средств клиенту (приходная транзакция), ```-``` — списание средств (расходная транзакция);
 - ```term_id``` — идентификатор терминала;


#### Таблица ```gender_train.csv```

##### Описание
Данная таблица содержит информацию по полу для части клиентов, для которых он известен. Для остальных клиентов пол неизвестен.

##### Формат данных
```
customer_id,gender
111111,0
111112,1
...
```

##### Описание полей
 - ```customer_id``` — идентификатор клиента;
 - ```gender``` — пол клиента;

### Таблица ```tr_mcc_codes.csv```

##### Описание
Данная таблица содержит описание mcc-кодов транзакций.

##### Формат данных
```
mcc_code;mcc_description
1000;словесное описание mcc-кода 1000
2000;словесное описание mcc-кода 2000
...
```

##### Описание полей
 - ```mcc_code``` – mcc-код транзакции;
 - ```mcc_description``` — описание mcc-кода транзакции.

````python
import pandas as pd
import matplotlib.pyplot as plt

transactions_short = pd.read_csv('transactions_short.csv')
tr_mcc_codes = pd.read_csv('tr_mcc_codes.csv', sep=';')
gender_train = pd.read_csv('gender_train.csv')

print('transactions_short:', transactions_short.shape)
print(transactions_short.head(3))
print('\ntr_mcc_codes:', tr_mcc_codes.shape)
print(tr_mcc_codes.head(3))
print('\ngender_train:', gender_train.shape)
print(gender_train.head(3))
````

````python
# ── Разведочный анализ данных ──────────────────────────────────────────────

datasets = {
    'transactions_short': transactions_short,
    'tr_mcc_codes': tr_mcc_codes,
    'gender_train': gender_train,
}

for name, df in datasets.items():
    print(f'═══ {name} ═══')
    print(f'Размер: {df.shape[0]:,} строк × {df.shape[1]} столбцов')

    # Типы признаков
    num_cols = df.select_dtypes(include='number').columns.tolist()
    cat_cols = df.select_dtypes(exclude='number').columns.tolist()
    print(f'Числовые признаки  ({len(num_cols)}): {num_cols}')
    print(f'Категориальные     ({len(cat_cols)}): {cat_cols}')

    # Пропуски
    missing = df.isnull().sum()
    missing = missing[missing > 0]
    if missing.empty:
        print('Пропуски: нет')
    else:
        print('Пропуски:')
        for col, cnt in missing.items():
            print(f'  {col}: {cnt} ({cnt/len(df)*100:.2f}%)')

    # Дубликаты
    dupes = df.duplicated().sum()
    print(f'Дублирующихся строк: {dupes}')

    # Базовая статистика числовых столбцов
    if num_cols:
        print('Статистика числовых признаков:')
        print(df[num_cols].describe().round(2).to_string())

    print()
````

````python
# Задание 2 — извлечь час, подсчитать приходные и расходные по каждому часу
transactions_short['tr_hour'] = (
    transactions_short['tr_datetime']
    .str.split(' ').str[1]
    .str.split(':').str[0]
    .astype(int)
)

incoming_by_hour = (
    transactions_short[transactions_short['amount'] > 0]
    .groupby('tr_hour').size()
    .rename('incoming_count')
)
outgoing_by_hour = (
    transactions_short[transactions_short['amount'] < 0]
    .groupby('tr_hour').size()
    .rename('outgoing_count')
)

hourly = pd.concat([incoming_by_hour, outgoing_by_hour], axis=1).fillna(0).astype(int)
hourly.index.name = 'tr_hour'
print(hourly.to_string())
````

````python
# Задание 2 (вопрос) — затраты в ночное время (00:00–05:59), топ категории + гистограмма
night_expenses = transactions_short[
    (transactions_short['tr_hour'] < 6) & (transactions_short['amount'] < 0)
].copy()
night_expenses['amount_abs'] = night_expenses['amount'].abs()

night_mcc = (
    night_expenses.groupby('mcc_code')['amount_abs'].sum()
    .reset_index()
    .merge(tr_mcc_codes, on='mcc_code')
    .nlargest(15, 'amount_abs')
    .reset_index(drop=True)
)

plt.figure(figsize=(12, 7))
plt.barh(night_mcc['mcc_description'][::-1], night_mcc['amount_abs'][::-1], color='darkblue')
plt.xlabel('Сумма затрат')
plt.title('Категории с наибольшими затратами в ночное время (00–06 ч)')
plt.tight_layout()
plt.show()

print('\nТоп-15 категорий ночных затрат:')
print(night_mcc[['mcc_code', 'mcc_description', 'amount_abs']].to_string(index=False))
print('\nАномалии: если какая-то категория значительно превышает остальные — это аномалия.')
````

## Вопросы к заданию 1.4

В поле tr_datetime можно выделит день tr_day, в который произошла транзакция. Это первые 2 цифры, отделенные пробелом от остальных символов. Сколько всего дней в данных? Выведите гистограмму по количеству транзакций в день.

````python
# Задание 3 — 10 дней с максимальной разницей в тратах между мужчинами и женщинами
tr_gender = transactions_short.merge(gender_train, on='customer_id', how='inner')
tr_expenses = tr_gender[tr_gender['amount'] < 0].copy()
tr_expenses['amount_abs'] = tr_expenses['amount'].abs()

# Траты по дням и полу
daily_gender = (
    tr_expenses.groupby(['tr_day', 'gender'])['amount_abs'].sum()
    .unstack(fill_value=0)
)
# gender: 0 = женщина, 1 = мужчина
daily_gender.columns.name = None
if 0 not in daily_gender.columns:
    daily_gender[0] = 0
if 1 not in daily_gender.columns:
    daily_gender[1] = 0
daily_gender['diff'] = (daily_gender[1] - daily_gender[0]).abs()
daily_gender['dominant_gender'] = (daily_gender[1] > daily_gender[0]).astype(int)

top10_days = daily_gender.nlargest(10, 'diff').reset_index()

# Для каждого из 10 дней — топ-категория по тратам (без учёта пола)
result_rows = []
for _, row in top10_days.iterrows():
    day = row['tr_day']
    day_data = tr_expenses[tr_expenses['tr_day'] == day]
    top_mcc_code = day_data.groupby('mcc_code')['amount_abs'].sum().idxmax()
    mcc_desc_val = tr_mcc_codes[tr_mcc_codes['mcc_code'] == top_mcc_code]['mcc_description'].values
    mcc_desc_val = mcc_desc_val[0] if len(mcc_desc_val) > 0 else 'Неизвестно'
    result_rows.append({
        'tr_day': day,
        'mcc_code': top_mcc_code,
        'mcc_description': mcc_desc_val,
        'dominant_gender': int(row['dominant_gender'])
    })

result_df = pd.DataFrame(result_rows)
print(result_df.to_string(index=False))
````
